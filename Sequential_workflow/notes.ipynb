# Imprt the Modules

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

## Creating the State

```python
class TemparatureState(TypedDict):
    temp_celcius: float
    temp_fahrenheit: float
```

```python
def convert_temp(state: TemparatureState) -> TemparatureState:
    celcius = state['temp_celcius']
    fahrenheit = (celcius * 9/5) + 32
    state['temp_fahrenheit'] = round(fahrenheit, 2)

    return state
```

## Define and Compile Graph

```python
graph = StateGraph(TemparatureState)

graph.add_node('convert_temp', convert_temp)

# Add edges to the graph
graph.add_edge(START, 'convert_temp')
graph.add_edge('convert_temp', END)

# Compile the graph
work_flow = graph.compile()
```

## Execute the Graph

```python
initial_graph = {'temp_celcius' : 28.5}
fila_state = work_flow.invoke(initial_graph)
print(fila_state)
```

```python
from pydantic import BaseModel
class GreetNode(BaseModel):
    messages: str = ""
    isPassed: bool = True

def greet(state: GreetNode):
    "Display the greet message."
    print(state.messages)
    return state
```

```python
graph = StateGraph(GreetNode)
graph.add_node("greet", greet)

graph.add_edge(START, "greet")
graph.add_edge("greet", END)
```

```python
print(graph)
```

```python
final_graph = graph.compile()
```
